```python
# pandas helps us load and work with data tables
import pandas as pd

# numpy helps us handle missing values and do math on columns
import numpy as np

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"Numpy version: {np.__version__}")
```

```python
# Load the Raw Data
train = pd.read_csv('data/train.csv', 
                    low_memory=False, 
                    dtype={'StateHoliday': str})

store = pd.read_csv('data/store.csv', low_memory=False)

print("Files loaded successfully!")
print(f"Train shape: {train.shape}")
print(f"Store shape: {store.shape}")
print("Warning fixed — StateHoliday loaded as text!")
```

```python
print("MISSING VALUES IN TRAIN:")
missing_train = train.isnull().sum()
missing_percent_train = (train.isnull().sum() / len(train)) * 100
missing_summary_train = pd.DataFrame({
    'Missing Values': missing_train,
    'Percentage': missing_percent_train
})
print(missing_summary_train)
```

```python
print("MISSING VALUES IN STORE:")
missing_store = store.isnull().sum()
missing_percent_store = (store.isnull().sum() / len(store)) * 100
missing_summary_store = pd.DataFrame({
    'Missing Values': missing_store,
    'Percentage': missing_percent_store
})
print(missing_summary_store)
```

```python
# Fix 1 — CompetitionDistance
# Fill the 3 missing values with the median distance of all stores
# We use median instead of average because it is less affected by very large distances
store['CompetitionDistance'] = store['CompetitionDistance'].fillna(store['CompetitionDistance'].median())

# Fix 2 — CompetitionOpenSinceMonth and CompetitionOpenSinceYear
# Fill with 0 to indicate the opening date is unknown
store['CompetitionOpenSinceMonth'] = store['CompetitionOpenSinceMonth'].fillna(0)

store['CompetitionOpenSinceYear'] = store['CompetitionOpenSinceYear'].fillna(0)

# Fix 3 — Promo2 related columns
# Fill with 0 for numbers and "None" for text
store['Promo2SinceWeek'] = store['Promo2SinceWeek'].fillna(0)

store['Promo2SinceYear'] = store['Promo2SinceYear'].fillna(0)

store['PromoInterval'] = store['PromoInterval'].fillna("None")

print("Missing values fixed!")
```

```python
print("MISSING VALUES IN STORE AFTER FIXING:")
print(store.isnull().sum())
```

```python
train['Date'] = pd.to_datetime(train['Date'])

print("Date column converted successfully!")
print(f"New data type: {train['Date'].dtype}")
print(f"Earliest date: {train['Date'].min()}")
print(f"Latest date: {train['Date'].max()}")
```

```python
print("STORES CLOSED BUT SHOWING SALES:")
print(train[(train['Open'] == 0) & (train['Sales'] > 0)].shape[0])

print("\nSTORES OPEN BUT SHOWING ZERO SALES:")
print(train[(train['Open'] == 1) & (train['Sales'] == 0)].shape[0])

print("\nNEGATIVE SALES:")
print(train[train['Sales'] < 0].shape[0])

print("\nNEGATIVE CUSTOMERS:")
print(train[train['Customers'] < 0].shape[0])
```

```python
rows_before = len(train)

train = train[(train['Open'] == 0) | (train['Sales'] > 0)]

rows_after = len(train)

print(f"Rows before: {rows_before}")
print(f"Rows removed: {rows_before - rows_after}")
print(f"Rows after: {rows_after}")
```

```python
print("BASIC STATISTICS FOR TRAIN:")
train[['Sales', 'Customers']].describe()
```

```python
print("BASIC STATISTICS FOR STORE:")
store[['CompetitionDistance']].describe()
```

```python
df = train.merge(store, on='Store', how='left')

print(f"Merged dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

```python
df.to_csv('data/cleaned_data.csv', index=False)

print("Clean dataset saved successfully!")
print(f"File saved to: data/cleaned_data.csv")
print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")
```
