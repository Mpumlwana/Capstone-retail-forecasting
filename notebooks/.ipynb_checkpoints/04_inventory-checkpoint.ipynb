```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
plt.rcParams["figure.figsize"] = (12, 5)

print("Libraries imported successfully!")
```

```python
xgb_model = joblib.load('../models/xgboost_model.pkl')
le = joblib.load('../models/label_encoder.pkl')

df = pd.read_csv('../data/cleaned_data.csv', low_memory=False)
df['Date'] = pd.to_datetime(df['Date'])

print("Model loaded successfully!")
print("Dataset loaded successfully!")
print(f"Dataset shape: {df.shape}")
```

```python
import os
print(os.getcwd())
print(os.listdir('../models'))
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['DayOfMonth'] = df['Date'].dt.day
df['WeekOfYear'] = df['Date'].dt.isocalendar().week.astype(int)

storetype_map = {'a': 0, 'b': 1, 'c': 2, 'd': 3}
assortment_map = {'a': 0, 'b': 1, 'c': 2}

df['StoreType_encoded'] = df['StoreType'].map(storetype_map)
df['Assortment_encoded'] = df['Assortment'].map(assortment_map)

features = [
    'Store', 'DayOfWeek', 'Year', 'Month', 'DayOfMonth',
    'WeekOfYear', 'Promo', 'SchoolHoliday',
    'StoreType_encoded', 'Assortment_encoded',
    'CompetitionDistance', 'Promo2'
]

print("Features recreated successfully!")
print(f"StoreType mapping: {storetype_map}")
print(f"Assortment mapping: {assortment_map}")
```

```python
last_date = df['Date'].max()
future_dates = pd.date_range(
    start=last_date + pd.Timedelta(days=1),
    periods=30,
    freq='D'
)

print(f"Last date in dataset: {last_date.date()}")
print(f"Forecasting from: {future_dates[0].date()}")
print(f"Forecasting to: {future_dates[-1].date()}")
print(f"Total days to forecast: {len(future_dates)}")
```

```python
store_info = df.groupby('Store').agg({
    'StoreType_encoded': 'first',
    'Assortment_encoded': 'first',
    'CompetitionDistance': 'first',
    'Promo2': 'first'
}).reset_index()

future_rows = []

for date in future_dates:
    for _, store in store_info.iterrows():
        future_rows.append({
            'Store': int(store['Store']),
            'DayOfWeek': date.dayofweek + 1,
            'Year': date.year,
            'Month': date.month,
            'DayOfMonth': date.day,
            'WeekOfYear': date.isocalendar()[1],
            'Promo': 0,
            'SchoolHoliday': 0,
            'StoreType_encoded': int(store['StoreType_encoded']),
            'Assortment_encoded': int(store['Assortment_encoded']),
            'CompetitionDistance': store['CompetitionDistance'],
            'Promo2': int(store['Promo2'])
        })

future_df = pd.DataFrame(future_rows)

print(f"Future feature table created!")
print(f"Shape: {future_df.shape}")
print(f"Expected: {1115 * 30} rows x 12 columns")
```

```python
X_future = future_df[features]
future_df['Predicted_Sales'] = xgb_model.predict(X_future)

future_df['Predicted_Sales'] = future_df['Predicted_Sales'].clip(lower=0)

print("Forecasts generated successfully!")
print(f"Total predictions: {len(future_df):,}")
print(f"\nSample predictions for Store 1:")
print(future_df[future_df['Store'] == 1][
    ['Store', 'DayOfWeek', 'Predicted_Sales']
].head(7).to_string(index=False))
```

```python
LEAD_TIME = 7
Z_SCORE = 1.65

inventory = future_df.groupby('Store').agg(
    Avg_Daily_Demand=('Predicted_Sales', 'mean'),
    Std_Daily_Demand=('Predicted_Sales', 'std'),
    Total_Forecasted_Sales=('Predicted_Sales', 'sum')
).reset_index()

inventory['Safety_Stock'] = (
    Z_SCORE * inventory['Std_Daily_Demand'] * np.sqrt(LEAD_TIME)
).round(0)

inventory['Reorder_Point'] = (
    inventory['Avg_Daily_Demand'] * LEAD_TIME + inventory['Safety_Stock']
).round(0)

inventory['Reorder_Point'] = inventory['Reorder_Point'].clip(lower=0)
inventory['Safety_Stock'] = inventory['Safety_Stock'].clip(lower=0)

print("Inventory metrics calculated!")
print(f"\nSample results for first 5 stores:")
print(inventory.head().to_string(index=False))
```

```python
inventory.to_csv('../data/inventory_recommendations.csv', index=False)

print("Inventory recommendations saved!")
print(f"File: data/inventory_recommendations.csv")
print(f"Stores covered: {len(inventory)}")
print(f"\nOverall summary:")
print(f"  Average reorder point across all stores: "
      f"€{inventory['Reorder_Point'].mean():,.0f}")
print(f"  Average safety stock across all stores:  "
      f"€{inventory['Safety_Stock'].mean():,.0f}")
print(f"  Highest reorder point (Store "
      f"{inventory.loc[inventory['Reorder_Point'].idxmax(), 'Store']:.0f}): "
      f"€{inventory['Reorder_Point'].max():,.0f}")
print(f"  Lowest reorder point (Store "
      f"{inventory.loc[inventory['Reorder_Point'].idxmin(), 'Store']:.0f}): "
      f"€{inventory['Reorder_Point'].min():,.0f}")
```

```python
store1_forecast = future_df[future_df['Store'] == 1].copy()
store1_forecast['Date'] = future_dates[:len(store1_forecast)]

store1_rop = inventory[inventory['Store'] == 1]['Reorder_Point'].values[0]
store1_safety = inventory[inventory['Store'] == 1]['Safety_Stock'].values[0]

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

ax1.bar(store1_forecast['Date'], 
        store1_forecast['Predicted_Sales'],
        color='steelblue', alpha=0.7)
ax1.set_title('Store 1 — Daily Predicted Sales (August 2015)', fontsize=13)
ax1.set_xlabel('Date')
ax1.set_ylabel('Predicted Sales (€)')
ax1.tick_params(axis='x', rotation=45)

store1_forecast['Cumulative_Sales'] = store1_forecast['Predicted_Sales'].cumsum()
ax2.plot(store1_forecast['Date'], 
         store1_forecast['Cumulative_Sales'],
         color='steelblue', linewidth=2, label='Cumulative Sales')
ax2.axhline(y=store1_rop, color='red', linestyle='--', 
            linewidth=1.5, label=f'Reorder Point (€{store1_rop:,.0f})')
ax2.axhline(y=store1_safety, color='orange', linestyle='--',
            linewidth=1.5, label=f'Safety Stock (€{store1_safety:,.0f})')
ax2.set_title('Store 1 — Cumulative Sales vs Reorder Point', fontsize=13)
ax2.set_xlabel('Date')
ax2.set_ylabel('Cumulative Sales (€)')
ax2.legend()
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('../outputs/08_inventory_store1.png', dpi=150)
plt.show()
print("Chart saved!")
```

```python
print("=" * 55)
print("  INVENTORY ENGINE COMPLETE — TASK 4 SUMMARY")
print("=" * 55)
print(f"\n  Forecast period:  1 August — 30 August 2015")
print(f"  Stores covered:   {len(inventory):,}")
print(f"  Total predictions:{len(future_df):,}")
print(f"  Lead time assumed: {LEAD_TIME} days")
print(f"  Service level:     95% (Z-score {Z_SCORE})")
print()
print("  INVENTORY METRICS:")
print(f"  Avg daily demand per store: "
      f"€{inventory['Avg_Daily_Demand'].mean():,.0f}")
print(f"  Avg safety stock per store: "
      f"€{inventory['Safety_Stock'].mean():,.0f}")
print(f"  Avg reorder point per store:"
      f"€{inventory['Reorder_Point'].mean():,.0f}")
print()
print("  SAVED FILES:")
print("  • data/inventory_recommendations.csv")
print("  • outputs/08_inventory_store1.png")
print("=" * 55)
```

```python
store1 = future_df[future_df['Store'] == 1].copy()
store1['Date'] = future_dates[:len(store1)]
daily_sales = store1['Predicted_Sales'].values

rop = inventory.loc[inventory['Store'] == 1, 'Reorder_Point'].values[0]
safety = inventory.loc[inventory['Store'] == 1, 'Safety_Stock'].values[0]
avg_demand = inventory.loc[inventory['Store'] == 1, 'Avg_Daily_Demand'].values[0]

order_qty = avg_demand * 14
stock = rop + avg_demand * 7
arrival_day = None
stock_levels = []
order_days = []

for day in range(len(daily_sales)):
    if arrival_day is not None and day == arrival_day:
        stock += order_qty
        arrival_day = None
    stock = max(stock - daily_sales[day], 0)
    if stock <= rop and arrival_day is None:
        arrival_day = day + LEAD_TIME
        order_days.append(day)
    stock_levels.append(stock)

plt.figure(figsize=(12, 5))
plt.plot(store1['Date'], stock_levels, color='steelblue', linewidth=2, label='Stock on hand')
plt.axhline(y=rop, color='red', linestyle='--', label=f'Reorder Point (€{rop:,.0f})')
plt.axhline(y=safety, color='orange', linestyle='--', label=f'Safety Stock (€{safety:,.0f})')
for d in order_days:
    plt.axvline(x=store1['Date'].iloc[d], color='green', linestyle=':', alpha=0.8)
plt.title('Store 1: Stock Level and Reorder Triggers (August 2015)', fontsize=13)
plt.xlabel('Date')
plt.ylabel('Stock value (€)')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../outputs/09_stock_simulation_store1.png', dpi=150)
plt.show()
print("Order placed on day(s):", [d + 1 for d in order_days])
```
